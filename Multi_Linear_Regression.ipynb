```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import  r2_score
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv("House Price.csv")
data
```

```python
x1=data.loc[1:20,'Area_sqft']
x1
```

```python
x2=data.loc[1:20,'Bedrooms']
x2
```

```python
X=np.column_stack((x1,x2))
X
```

```python
Y=data.loc[1:20,'Price_Lakhs']
Y
```

```python
model=LinearRegression()
model.fit(X,Y)
```

```python
Area_size=int(input("Enter your House area:"))
Bedroom=int(input("Enter your no. of bedrooms:"))
inputss=np.column_stack((Area_size,Bedroom))
print(model.predict(inputss))
```

```python
x_test=X
y_test=Y
y_pred=model.predict(x_test)
accuracy=r2_score(y_test,y_pred)
print(accuracy*100)
```

```python
import matplotlib.pyplot as plt
fig=plt.figure()
ax=fig.add_subplot(121,projection='3d')
ax.scatter(data.loc[:,'Area_sqft'],data.loc[:,'Bedrooms'],data.loc[:,'Price_Lakhs'])
plt.title("House requirements vs price")
ax.set_xlabel('Area')
ax.set_ylabel('No. of Bedrooms')
ax.set_zlabel('Price')
plt.show()
```

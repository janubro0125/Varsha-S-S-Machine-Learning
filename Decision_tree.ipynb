```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor
```

```python
data=pd.read_csv("Agent-Shield.csv")
data
```

```python
x=data[['latency_ms','tokens_used']]
x
```

```python
y=data['threat_category']
y
```

```python
from sklearn.preprocessing import LabelEncoder
threat_category=data['threat_category']
le=LabelEncoder()
y=le.fit_transform(threat_category)
y
```

```python
from sklearn.model_selection import  train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=40)
```

```python
model=DecisionTreeRegressor(criterion='squared_error')
model.fit(x_train,y_train)
```

```python
Latency=int(input("Enter your latency:"))
tokens_used=int(input("Enter your no.of tokens used:"))
x1=[[Latency,tokens_used]]
print(model.predict(x1))
```
